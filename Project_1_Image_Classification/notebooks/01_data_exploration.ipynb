5. Dataset loading / verification

```python
import os
from torchvision import datasets

train_dir = r"D:\deep learning project\data\intel image classification\seg_train\seg_train"
test_dir = r"D:\deep learning project\data\intel image classification\seg_test\seg_test"

train_dataset = datasets.ImageFolder(train_dir)
test_dataset = datasets.ImageFolder(test_dir)

print("Classes:", train_dataset.classes)
print("Training images:", len(train_dataset))
print("Test images:", len(test_dataset))
```

6. Sample images

```python
import matplotlib.pyplot as plt
from PIL import Image

fig, axes = plt.subplots(2, 3, figsize=(12, 7))

for class_index, class_name in enumerate(train_dataset.classes):
    # Find the first image belonging to this class
    image_index = train_dataset.targets.index(class_index)

    image_path, label = train_dataset.samples[image_index]
    image = Image.open(image_path).convert("RGB")

    row = class_index // 3
    col = class_index % 3

    axes[row, col].imshow(image)
    axes[row, col].set_title(class_name)
    axes[row, col].axis("off")

plt.tight_layout()
plt.show()
```

8. Preprocessing

```python
from torchvision import transforms
from torchvision import datasets

transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor()
])

full_dataset = datasets.ImageFolder(
    train_dir,
    transform=transform
)

test_dataset = datasets.ImageFolder(
    test_dir,
    transform=transform
)

print("Total training dataset:", len(full_dataset))
print("Test dataset:", len(test_dataset))
```

9. Train / Validation Split

```python
from torch.utils.data import random_split

# 80% training, 20% validation
train_size = int(0.8 * len(full_dataset))
val_size = len(full_dataset) - train_size

train_dataset, val_dataset = random_split(
    full_dataset,
    [train_size, val_size],
    generator=torch.Generator().manual_seed(42)
)

print("Training images:", len(train_dataset))
print("Validation images:", len(val_dataset))
print("Test images:", len(test_dataset))
```

10. DataLoaders

```python
from torch.utils.data import DataLoader

BATCH_SIZE = 32

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=0
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0
)

print("Training batches:", len(train_loader))
print("Validation batches:", len(val_loader))
print("Test batches:", len(test_loader))
```

11. Batch Verification

```python
images, labels = next(iter(train_loader))

print("Image batch shape:", images.shape)
print("Labels shape:", labels.shape)
print("Image data type:", images.dtype)
print("Minimum pixel value:", images.min().item())
print("Maximum pixel value:", images.max().item())
```
